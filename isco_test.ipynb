# Tests of `isco.py`

Checks of the ISCO radius and frequency against the analytic Kerr limits and against
Table 1 of arXiv:2012.01350.
Prints the cutoff frequency $2 f_{\rm ISCO}$ of the remnant for the
configurations of Figs. 2-3.

Run from the repository root.

```python
import numpy as np
from isco import compute_r_isco, compute_f_isco, isco_cutoff
```

```python
# Reference configuration
m1, m2 = 24.0, 8.0   # q = 3
chi1, chi2 = 0.7, 0.6
```

```python
# Run the functions for a given configuration
M = (m1 + m2)
q = m1 / m2
f_cutoff, chi_f, r_isco = isco_cutoff(M, q, chi1, chi2)
print(f"chi_f = {chi_f:.4f}")
print(f"r_isco/M = {r_isco/M:.4f}")
print(f"f_cutoff = {f_cutoff:.4f}")
```

```python
# Tests for chi_f, r_isco and f_cutoff

M = 1.0

print("Schwarzschild (chi_f = 0)")
r = compute_r_isco(0.0, M)
f = compute_f_isco(M, 0.0, r)
print(f"r_isco = {r:.6f} M   (expected: 6.0 M)")
print(f"M*f_isco = {M*f:.6f}  (expected: {6**-1.5/(2*np.pi):.6f})")
print()

print("prograde extremal Kerr (chi_f -> 1)")
for chi_f in [0.9, 0.99, 0.999, 0.9999]:
    r = compute_r_isco(chi_f, M)
    print(f"chi_f={chi_f:7.4f}  r_isco = {r:.6f} M  (expected: -> 1 M)")
print()

print("retrograde extremal Kerr (chi_f -> -1)")
for chi_f in [-0.9, -0.99, -0.999, -0.9999]:
    r = compute_r_isco(chi_f, M)
    print(f"chi_f={chi_f:7.4f}  r_isco = {r:.6f} M  (expected -> 9 M)")
print()


M_tot = 32.0

print("Aligned spins (chi_f > 0)")
f_cutoff, chi_f, r_isco = isco_cutoff(M_tot, q=3, chi_1=0.7, chi_2=0.6)
print(f"chi_f = {chi_f:.4f}  (expected to be > 0)")
print(f"r_isco = {r_isco:.4f} M_sun  (r/M = {r_isco/M_tot:.4f})")
print(f"f_cutoff = {f_cutoff:.2f} Hz")
print()

print("Anti-aligned spins (chi_f < 0)")
f_cutoff, chi_f, r_isco = isco_cutoff(M_tot, q=10, chi_1=-0.7, chi_2=-0.6)
print(f"chi_f = {chi_f:.4f}  (expected to be < 0)")
print(f"r_isco = {r_isco:.4f} M_sun  (r/M = {r_isco/M_tot:.4f})")
print(f"f_cutoff = {f_cutoff:.2f} Hz")
```

```python
# Cross-check against Table 1 of https://arxiv.org/pdf/2012.01350 (page 19)

M = 1.0
for chi_f in [-1, -0.9, -0.5, 0, 0.5, 0.9, 1]:
    r = compute_r_isco(chi_f, M)
    v = (compute_f_isco(M, chi_f, r) * 2 * np.pi * M) ** (1 / 3)
    print(f"chi_f = {chi_f:.1f}")
    print(f"r_isco/M = {r/M:.9f}")
    print(f"v_isco = {v:.10f}")
    print()
```

```python
# ISCO cutoff for all the configurations of Figs. 2-3

from config.params import (M_TOTAL_SIM, Q_MASS_RATIO, Q_MASS_RATIO_CONFIGS_FIG3,
                           CHI_1, CHI_2, SPIN_CONFIGS_FIG2)

configs = [(M_TOTAL_SIM, Q_MASS_RATIO, c1, c2) for c1, c2 in SPIN_CONFIGS_FIG2] \
        + [(M_TOTAL_SIM, q, CHI_1, CHI_2) for q in Q_MASS_RATIO_CONFIGS_FIG3]

for M, q, c1, c2 in configs:
    f_cut, chi_f, _ = isco_cutoff(M, q, c1, c2)
    print(f"q = {q:4.2f}, chi = ({c1}, {c2}):  chi_f = {chi_f:.3f},  f_cut (ISCO) = {f_cut:6.1f} Hz")
```
