# Tests of `tidal_heating_correction.py`

Checks of the coefficients $\Psi_5$, $\Psi_8$ and of the phase correction $\delta\psi$ in the
non-spinning case, in the GR limit ($q^{\rm BH} = 0$, Kerr tidal heating), for a positive charge
and in the case without horizon (complex $\Psi_8$).

In the GR limit, $\delta\psi$ was also compared with the tidal heating phase of the code of
arXiv:2311.17554 (`psiTH` in [gwfit](https://git.ligo.org/samanwaya.mukherjee/gwfit), LIGO GitLab; not distributed here): the two agree to machine precision
($\lesssim 10^{-15}$ rad over the whole band).

Run from the repository root.

```python
import numpy as np
from tidal_heating_correction import A5_single, A8_single, Psi5_and_Psi8, delta_psi_tidal
```

```python
freqs = np.array([10., 20., 50., 100.])

# Parameters
m_1, m_2 = 24.0, 8.0   # q=3, M_tot=32 Msun
chi_1, chi_2 = 0.7, 0.6
```

```python
# Tests for Ψ5 and Ψ8

# chi = 0, q_BH = 0
Psi_5, Psi_8 = Psi5_and_Psi8(m_1, m_2, 0.0, 0.0, 0.0)
print(f"\nPsi5 = {Psi_5:.6f}   (expected 0.0)")
print(f"Psi8 = {Psi_8:.6f}   (expected > 0)")
delta_psi_nospin = delta_psi_tidal(freqs, m_1, m_2, 0.0, 0.0, 0.0)
print("delta_psi (chi=0, q_BH=0):")
print(delta_psi_nospin)

# chi = 0, q_BH = 0.5
Psi_5, Psi_8 = Psi5_and_Psi8(m_1, m_2, 0.0, 0.0, 0.5)
print(f"\nPsi5 = {Psi_5:.6f}   (expected 0.0)")
print(f"Psi8 = {Psi_8:.6f}   (expected > 0)")
delta_psi_charged_nospin = delta_psi_tidal(freqs, m_1, m_2, 0.0, 0.0, 0.5)
print("delta_psi (chi=0, q_BH=0.5):")
print(delta_psi_charged_nospin)
```

```python
# Tests for delta_psi (difference between GR and non-GR)

# q_BH = 0
q_BH_GR = 0.0 # Kerr tidal heating has to be recovered
Psi_5_gr, Psi_8_gr = Psi5_and_Psi8(m_1, m_2, chi_1, chi_2, q_BH_GR)
print(f"GR limit (q_BH=0):  Ψ5 = {Psi_5_gr:.6f},  Ψ8 = {Psi_8_gr:.6f}")
delta_psi_gr = delta_psi_tidal(freqs, m_1, m_2, chi_1, chi_2, q_BH_GR)
print("delta_psi (q_BH=0):")
print(delta_psi_gr)

# q_BH = 0.5
q_BH_test = 0.5
Psi_5_c, Psi_8_c = Psi5_and_Psi8(m_1, m_2, chi_1, chi_2, q_BH_test)
print(f"q_BH={q_BH_test}:           Ψ5 = {Psi_5_c:.6f},  Ψ8 = {Psi_8_c:.6f}")
delta_psi_test = delta_psi_tidal(freqs, m_1, m_2, chi_1, chi_2, q_BH_test)
print("delta_psi (q_BH=0.5):")
print(delta_psi_test)
```

```python
# No horizon (naked singularity), when 1 - chi_i^2 + q_BH (M/m_i)^2 < 0

Psi_5, Psi_8 = Psi5_and_Psi8(m_1, m_2, chi_1, chi_2, -0.7)
print(f"Psi5 = {Psi_5:.6f}")
print(f"Psi8 = {Psi_8}")
print(f"Psi8 is complex: {np.iscomplex(Psi_8)}")

delta_psi = delta_psi_tidal(freqs, m_1, m_2, chi_1, chi_2, -0.7)
print(f"delta_psi = {delta_psi}")
```
